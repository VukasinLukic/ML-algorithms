```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
df.info()
```

```python
from sklearn.model_selection import train_test_split
X=df.drop(columns='Left')
y=df['Left']
```

```python
X_scaled=pd.DataFrame(data=scaler.fit_transform(X),columns=X.columns)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42,stratify=y)
```

```python
X_train
```

```python
#Arhitektura nase neuornske mreze je 1 ulaz sa 9 neurona(9 feature kolona) , 2 skrivena layera sa 64 i 32 neuorona , i jedan izlaz sa 1 neuronom . 9-64-32-1
#Neuronska mreza se sastoji od layera neurona , gde je svaki layer povezan sa prethodnim nivom. Prvi proces koji se desava se zove ForwardPropaganation kada se pozove metoda fit , tada ulazni podaci se prosledjuju do prvog skrivenog nivoa neurona , gde im se dodaju tezine (kao cene na granama) i biasi biasi omogucavaju fleskbilnost u Neuronskoj mrezi . KAda se sracunaju bias i tezina , na njih deluje aktivaciona fja Relu , koja te ulazne podatke ako su negativni svede na nulu a ako nisu ostavi kakvi jesu onda ulazi nastave dalje ka 2 sloju , potom dolaze do izlaznog sloja u kome se sracunavaju greske od stvarne vrednosti iz ytraina koju trebaju da imaju , tada se dobija loss koji Neurosnka mreza i cilja da smanji svakom iteracijom , i onda se desava zapravo sustina NeurosnkeMreze BACKPROPAGATION proces, kada se sracunaju te greske one se salju nazad do ulaza gde im se koriguju weights i biasi , tako da u svakoj sledecoj iteraciji su te greske sve manje i manje , ovo se ponavlja dok se ne dodje do zadovoljene tacnosti ili max iteriacija definasen u mlp classifareu .
```

```python
from sklearn.neural_network import MLPClassifier
clf = MLPClassifier(hidden_layer_sizes=(64,32),activation="relu",random_state=42, max_iter=10000,solver="sgd",verbose=1).fit(X_train, y_train)
```

```python
plt.plot(clf.loss_curve_)
```

```python
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix, ConfusionMatrixDisplay, precision_recall_fscore_support

y_pred = clf.predict(X_test)
ConfusionMatrixDisplay.from_predictions(
   y_test, y_pred,cmap="ocean")
```

```python
help(precision_recall_fscore_support)

#Po definiciji matrica konfuzije je metrika klasfikiacije ,C je matrica oblika C(i,j). sa 4 vrednosti Tn,Fp,Fn,Tp
```

```python
print(classification_report(y_test, y_pred))
```

```python
# The precision is the ratio ``tp / (tp + fp)`` where ``tp`` is the number of
#     true positives and ``fp`` the number of false positives. The precision is
#     intuitively the ability of the classifier not to label a negative sample as
#     positive.

# The recall is the ratio ``tp / (tp + fn)`` where ``tp`` is the number of
#     true positives and ``fn`` the number of false negatives. The recall is
#     intuitively the ability of the classifier to find all the positive samples.
#
#     The F-beta score can be interpreted as a weighted harmonic mean of
#     the precision and recall, where an F-beta score reaches its best
#     value at 1 and worst score at 0.
```

```python

tn, fp, fn, tp = confusion_matrix(y_test, y_pred).ravel().tolist()
```
